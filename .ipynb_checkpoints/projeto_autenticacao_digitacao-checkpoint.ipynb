## 1. O problema

### 1.1 Contexto
A senha ainda é o principal mecanismo de autenticação em aplicativos de banco, e-mail e redes sociais — inclusive no smartphone. O problema é que **a senha só verifica *o que* a pessoa sabe, não *quem* ela é**:

- Senhas vazam em ataques a empresas e são vendidas em listas;
- Muitas pessoas reutilizam a mesma senha em vários serviços;
- Golpes de engenharia social (phishing, falsas centrais de atendimento) levam a vítima a entregar a senha;
- Em celulares roubados, o criminoso muitas vezes já tem acesso ao aparelho desbloqueado.

> 📌 *Sugestão: inclua aqui um dado recente e com fonte sobre fraudes digitais no Brasil (por exemplo, relatórios da FEBRABAN ou do Fórum Brasileiro de Segurança Pública sobre golpes e roubo de celulares). Um número concreto fortalece muito a justificativa.*

### 1.2 A oportunidade: biometria comportamental
Cada pessoa tem um **ritmo próprio de digitação**: quanto tempo segura cada tecla, quanto demora entre uma tecla e outra, onde hesita. Esse padrão é chamado de **dinâmica de digitação** (*keystroke dynamics*).

Diferente do reconhecimento facial ou da impressão digital, essa verificação:
- **é invisível** para o usuário (não exige nenhum passo extra);
- **não precisa de hardware especial** (qualquer teclado ou tela sensível ao toque serve);
- **não armazena imagem do rosto** nem outro dado biométrico físico;
- funciona como **segunda camada de segurança**: mesmo que a senha tenha vazado, o impostor não digita como o dono.

### 1.3 Por que usar dados e Machine Learning?
Não existe uma regra simples do tipo *"se a pessoa demorar mais de 0,2 s na letra X, é impostor"*. O padrão de cada pessoa é formado por **dezenas de tempos combinados**, que variam um pouco a cada tentativa. Modelos de Machine Learning conseguem aprender essas combinações a partir de exemplos e separar o usuário legítimo de outras pessoas, algo inviável de fazer manualmente em escala.

### 1.4 Objetivo do projeto
Construir e avaliar um modelo de classificação que, a partir dos tempos de digitação de uma senha, indique se a digitação é do **usuário legítimo (genuíno)** ou de **outra pessoa (impostor)**, comparando **Árvore de Decisão** e **XGBoost**.

**Como vamos medir o sucesso** (métricas usuais em biometria):
| Métrica | Significado | Por que importa |
|---|---|---|
| **FAR** — *False Acceptance Rate* | % de impostores aceitos | Falha de **segurança** |
| **FRR** — *False Rejection Rate* | % de usuários legítimos bloqueados | Falha de **experiência do usuário** |
| **EER** — *Equal Error Rate* | ponto em que FAR = FRR | Resume o modelo em um único número |
| **AUC-ROC** | capacidade geral de separar as classes | Comparação entre modelos |

### 1.5 Hipóteses iniciais (a investigar na análise exploratória)
- **H1:** Cada pessoa tem tempos de digitação suficientemente distintos para ser diferenciada das demais.
- **H2:** Com a prática (sessões mais tardias), a pessoa digita mais rápido e de forma mais consistente.
- **H3:** Trechos "difíceis" da senha (troca para maiúscula com *Shift*, passagem de letra para número) são os que mais diferenciam as pessoas.
- **H4:** O tempo que a tecla fica pressionada (*hold*) é mais estável por pessoa do que o intervalo entre teclas.

## 2. Fontes de dados

### 2.1 Base principal: CMU Keystroke Dynamics Benchmark Dataset
| Item | Descrição |
|---|---|
| **Origem** | Carnegie Mellon University (CMU) — Killourhy, K. S. & Maxion, R. A. |
| **Publicação** | *"Comparing Anomaly-Detection Algorithms for Keystroke Dynamics"*, IEEE/IFIP DSN 2009 |
| **Acesso** | Público e gratuito: https://www.cs.cmu.edu/~keystroke/ |
| **Arquivo** | `DSL-StrongPasswordData.csv` |
| **Conteúdo** | 51 voluntários digitaram a mesma senha **`.tie5Roanl`** 400 vezes cada (8 sessões × 50 repetições) |
| **Tamanho** | 20.400 linhas × 34 colunas |

**Por que esta base é adequada:**
- **Pública e não confidencial:** os participantes são identificados apenas por códigos (`s002`, `s003`...), sem nome ou qualquer dado pessoal; não há a senha real de ninguém (todos digitaram a mesma senha de teste).
- **Controlada:** todos digitaram a mesma senha, então as diferenças vêm da pessoa, não do texto.
- **Referência acadêmica:** é um *benchmark* muito usado em pesquisas, o que permite comparar nossos resultados com a literatura.
- **Senha realista:** `.tie5Roanl` mistura pontuação, letras, número e maiúscula, como as senhas "fortes" exigidas por bancos.

## 3. Coleta e primeira inspeção dos dados

### 3.1 Bibliotecas

```python
from pathlib import Path

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", 40)
sns.set_theme(style="whitegrid")
```

### 3.2 Coleta
O arquivo é baixado direto do site da CMU na primeira execução e salvo na pasta `data/`. Nas execuções seguintes, o notebook usa a cópia local.

```python
URL = "https://www.cs.cmu.edu/~keystroke/DSL-StrongPasswordData.csv"
ARQUIVO = Path("data") / "DSL-StrongPasswordData.csv"

if not ARQUIVO.exists():
    ARQUIVO.parent.mkdir(exist_ok=True)
    pd.read_csv(URL).to_csv(ARQUIVO, index=False)

df = pd.read_csv(ARQUIVO)
print(f"{df.shape[0]:,} linhas × {df.shape[1]} colunas".replace(",", "."))
df.head()
```

### 3.3 Entendendo as colunas
Cada linha é **uma digitação completa da senha** `.tie5Roanl` + `Enter`.

| Coluna | Significado |
|---|---|
| `subject` | Código do participante (quem digitou) |
| `sessionIndex` | Sessão de coleta (1 a 8) |
| `rep` | Repetição dentro da sessão (1 a 50) |
| `H.<tecla>` | **Hold** — tempo (s) que a tecla ficou pressionada |
| `DD.<tecla1>.<tecla2>` | **Down-Down** — tempo entre apertar a tecla 1 e apertar a tecla 2 |
| `UD.<tecla1>.<tecla2>` | **Up-Down** — tempo entre soltar a tecla 1 e apertar a tecla 2 (pode ser negativo se a pessoa aperta a próxima antes de soltar a anterior) |

Note que `DD = H + UD` para cada par de teclas.

```python
colunas_tempo = df.columns[3:]
resumo_colunas = pd.DataFrame({
    "tipo": [c.split(".")[0] for c in colunas_tempo],
    "teclas": [" → ".join(c.split(".")[1:]) for c in colunas_tempo],
}, index=colunas_tempo)
resumo_colunas["tipo"].value_counts().rename({"H": "Hold", "DD": "Down-Down", "UD": "Up-Down"})
```

### 3.4 Qualidade dos dados

```python
print("Tipos de dados:")
print(df.dtypes.value_counts(), "\n")
print("Valores nulos no total:", df.isna().sum().sum())
print("Linhas duplicadas:", df.duplicated().sum())
```

```python
print("Participantes:", df["subject"].nunique())
print("Sessões por participante:", df.groupby("subject")["sessionIndex"].nunique().unique())
print("Digitações por participante:", df["subject"].value_counts().unique())
```

✅ A base está **completa e balanceada**: nenhum valor nulo e exatamente 400 digitações para cada um dos 51 participantes.

### 3.5 Estatísticas descritivas dos tempos

```python
df[colunas_tempo].describe().T.round(3)
```

**Observações:**
- Os tempos típicos são de décimos de segundo (mediana do *hold* ≈ 0,08 s).
- Existem **valores extremos**: alguns intervalos passam de 10 s, provavelmente pausas ou distrações durante a coleta. Eles serão tratados na análise exploratória.
- Valores **negativos** em `UD` são esperados: indicam que a pessoa apertou a próxima tecla antes de soltar a anterior (digitação "sobreposta", típica de quem digita rápido).

### 3.6 Uma primeira olhada: cada pessoa tem seu ritmo?
Como prévia da análise exploratória, vamos calcular o **tempo total** para digitar a senha (soma dos intervalos Down-Down + o *hold* da última tecla) e comparar os participantes.

```python
cols_dd = [c for c in colunas_tempo if c.startswith("DD.")]
df["tempo_total"] = df[cols_dd].sum(axis=1) + df["H.Return"]

ordem = df.groupby("subject")["tempo_total"].median().sort_values().index

fig, ax = plt.subplots(figsize=(14, 5))
sns.boxplot(data=df, x="subject", y="tempo_total", order=ordem,
            showfliers=False, color="#7fb3d5", ax=ax)
ax.set_title("Tempo total para digitar a senha, por participante (ordenado pela mediana)")
ax.set_xlabel("Participante")
ax.set_ylabel("Tempo total (s)")
ax.tick_params(axis="x", rotation=90, labelsize=8)
plt.tight_layout()
plt.show()
```

**Leitura inicial:** os participantes têm velocidades bem diferentes entre si: o mais rápido tem mediana de cerca de 1,4 s, enquanto o mais lento passa de 4,5 s (mais que o triplo). Isso é um primeiro indício a favor da **H1**. Mas o tempo total sozinho não basta: várias pessoas têm velocidades parecidas. Por isso o modelo vai precisar das **31 medidas de tempo** combinadas.

Plano:
